# 环境搭建

本章使用 [uv](https://docs.astral.sh/uv/) 从零搭建 LangGraph 学习环境：

- **uv**：极快的 Python 包管理器（替代 pip/poetry），自动管理虚拟环境和依赖版本
- **Jupyter**：交互式 Notebook，本仓库所有实验都基于它运行
- **LangGraph**：构建有状态、多智能体 LLM 应用的框架

## 1. 安装 uv

uv 是一个独立可执行文件，不需要先装 Python（它本身可以下载管理 Python 版本）。

**Windows（PowerShell）：**

```powershell
winget install --id=astral-sh.uv -e
```

**macOS / Linux：**

```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
```

安装后验证：

````python
!uv --version
````

## 2. 初始化项目（空文件夹开始）

进入空文件夹，执行：

```bash
uv init --bare
```

它会产生：

- `pyproject.toml`：项目的依赖清单（类似 package.json）
- `.python-version`：锁定使用的 Python 版本
- `.venv/`：**虚拟环境**（独立于系统的 Python，依赖都装在这里）
- `uv.lock`：精确锁定的依赖版本（团队协作时保证一致）

> `--bare` 表示只生成最小骨架，不创建示例代码文件。

## 3. 安装依赖（自动创建虚拟环境）

`uv add` 会自动创建/复用 `.venv`，并把依赖写进 `pyproject.toml`：

```bash
# LangGraph 框架
uv add langgraph

# Jupyter Notebook 环境
uv add jupyterlab ipykernel

# （可选）按你使用的模型供应商添加：
# uv add langchain-anthropic    # Claude
# uv add langchain-openai       # OpenAI 兼容接口
# uv add langchain-ollama       # 本地模型

# （可选）画图工具
# uv add langgraph graphviz
```

## 4. 运行 Python / 启动 Jupyter

uv 有两种用法：

### 方式一：`uv run`（推荐，无需激活）

任何 `python`、`jupyter` 命令前加 `uv run`，会自动在 `.venv` 里执行，不用关心环境激活：

```bash
uv run python xxx.py
uv run jupyter lab
```

### 方式二：先激活虚拟环境

**Windows（PowerShell）：**

```powershell
Set-ExecutionPolicy -Scope Process -ExecutionPolicy RemoteSigned   
.venv\Scripts\Activate.ps1
```

**macOS / Linux：**

```bash
source .venv/bin/activate
```

激活后命令行的提示符会变成 `(.venv)`，此时直接敲 `python` / `jupyter` 即可。

> 本教程统一使用方式一（`uv run`），少一步、不容易忘。

## 5. Jupyter 使用要点

### 启动

```bash
uv run jupyter lab
```

会自动打开浏览器（默认 http://localhost:8888）。用 **JupyterLab** 而不用旧版 Notebook，界面更现代。

### 基本操作

- **快捷键**：`Shift+Enter` 运行当前单元格；`Esc` 退出编辑，`M` 转 Markdown 单元格，`Y` 转代码单元格
- **运行 Shell 命令**：在代码单元格里以 `!` 开头，如 `!uv --version`（本章上面就是这么用的）
- **查看依赖环境**：代码单元格里执行下面两行，确认 Jupyter 用的就是本项目环境

### 换一台机器恢复环境

克隆仓库后，只需一条命令即可装回所有依赖：

```bash
uv sync
```

uv 会根据 `uv.lock` 安装完全一致的版本，然后 `uv run jupyter lab` 直接可用。

## 常用命令速查

| 命令 | 作用 |
|---|---|
| `uv init --bare` | 初始化空项目 |
| `uv add <包名>` | 安装依赖并写入 pyproject.toml |
| `uv remove <包名>` | 卸载依赖 |
| `uv sync` | 按 lock 文件恢复全部依赖 |
| `uv run <命令>` | 在虚拟环境中运行命令 |
| `uv python install 3.12` | 安装指定版本的 Python |
| `uv tree` | 查看依赖树 |
| `uv pip list` | 查看已安装包（兼容 pip 语法） |
