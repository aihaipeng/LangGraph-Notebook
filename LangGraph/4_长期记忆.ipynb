```python
import os
from dotenv import load_dotenv
from langgraph.store.postgres import PostgresStore
import uuid

load_dotenv()

DB_URL = os.getenv("DATABASE_URL")


def ns(user_id: str):
    return ("memories", user_id)


with PostgresStore.from_conn_string(DB_URL) as store:  # type: ignore
    store.setup()

    store.put(ns("aihaipeng"), str(uuid.uuid4()), {"hobby": "编程", "age": 26})
    store.put(ns("huwenxin"), str(uuid.uuid4()), {"hobby": "旅游", "age": 26})
    store.put(ns("zhaohuatian"), str(uuid.uuid4()), {"hobby": "打游戏", "age": 18})

    # 查询 memories 命名空间下的所有数据
    for item in store.search(("memories",)):
        print(item)

    # 查询 memories 命名空间下的 aihaipeng 用户数据
    for item in store.search(("memories","aihaipeng")):
        print(item)
```
